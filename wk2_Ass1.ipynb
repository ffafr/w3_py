Filter only negative and zero in the list using list comprehension

numbers = [-4, -3, -2, -1, 0, 2, 4, 6]

```python
numbers = [-4, -3, -2, -1, 0, 4, 6]
new_numbers = [x for x in numbers if x <= 0 ]
print (new_numbers)
```
